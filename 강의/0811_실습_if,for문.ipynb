```python
order_database = [
    {"name": "김철수", "completed": True, "price": 60000},
    {"name": "이영희", "completed": False, "price": 35000},
    {"name": "박민수", "completed": True, "price": 25000},
    {"name": "최지영", "completed": True, "price": 85000},
    {"name": "정성민", "completed": False, "price": 90000}
]
total_paid = 0
order_cnt = 0
for i in order_database:
    if i["completed"]:
        if i['price'] >= 50000:
            order_cnt += 1
            total_paid += i['price']


print(f"우수 고객 주문 건수 :{order_cnt}")
print(f"우수 고객 총 결제 금액 :{total_paid}")
```

```python
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]
max_sweetness = 0
min_sweetness = 100
for i in fruit_sweetness:
    if i > max_sweetness:
        max_sweetness = i
    if i < min_sweetness:
        min_sweetness = i
print(f"최고 과일 당도:{max_sweetness}")
print(f"최저 과일 당도:{min_sweetness}")
f"과일 평균 당도: {sum(fruit_sweetness) / len(fruit_sweetness)}"
```

```python
import random
n = random.randint(1,51)
cnt = 0
while True:
    cnt += 1
    try:
        user_input = int(input("보물은 어디에 있을까? ( 1부터 50까지 숫자 입력 )"))
        if user_input < n:
            print("보물은 더 큰 숫자의 방에 있습니다.")
        elif user_input > n:
            print("보물은더 작은 방의 숫자에 있습니다")
        else:
            print(f"축하합니다! {cnt}번 만에 찾았어요.")
            break
    except ValueError:
                print("숫자만 적어주세요~")

        
```
