```python
menu = ["짜장면", "짬뽕", "탕수육"]

try:
    user_input = int(input("1~3번 메뉴중에 숫자로 메뉴를 선택해주세요 ! "))

except ValueError:
    print("숫자가 아니에요.")
    print("다시 입력해 주세요")
else:
    try:
        print(f"{menu[user_input - 1]} 금방 가져다 드리겟습니다 ! ")
    except IndexError:
        print("존재하지 않는 선택지에요.")
        print("다시 입력해 주세요")
finally:
    print("또 도움이 필요하시다면 벨을 눌러주세요.")
```

```python
def order_cnt(n):
    try:
        if n <= 0:
            raise ValueError
        return f"수량이 설정되었습니다. 수량 : {n} 개"
    except ValueError:
        return "수량이 음수입니다. 다시한번 확인 해주세요."



print(order_cnt(-10))
print(order_cnt(150))


```

```python
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

```python
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```

```python
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```
