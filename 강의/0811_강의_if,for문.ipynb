```python
num = 10
if num > 5:                     # num이 5보다 크다면 
    print("5보다 크다")
else:                           # num이 5보다 작다면 (첫번째 if의 조건에 맞지 않는 모든 경우)
    print("5보단 작다")
```

```python
age = int(input())

if len(str(age)) > 3:             # int의 len을 구할수는 없다고 함 .
    print("장난치지 마세요")
elif age > 19:                    # 집단의 범위가 넓은 조건부터 위쪽으로 올려야 한다.
    print("어른")
elif age > 16:
    print("고등학생")
elif age > 13:
    print("중학생")
elif age > 7:
    print("초등학생")
else:
    print("완전 꼬맹이")
```

```python
result = 0
for i in range(1,101):      # 1부터 100까지 세기
    if i % 2 != 0:          # i가 홀수면 result에 담기 // i 를 2로 나눳을떄 나머지가 0이 아닌녀석. 즉 홀수인 녀석
        result += i
result
```

```python
print("도서관 책 검색 시스템")
이달의_책 = "어린왕자"

while True:
    try: 
        user_input = input("1 : 책 검색,2 : 이달의 책 3 : 기부하기 4 : 종료")
        if int(user_input) < 1 or int(user_input) > 4:
            raise ValueError 
        if int(user_input) == 1:
            print("어떤 책을 찾으실지 몰라 다 준비해 봣어요")
        elif int(user_input) == 2:
            print(f"이달의 책은 {이달의_책} 입니다")
        elif int(user_input) == 3:
            donation = input("얼마나 기부하시려고요?")
            print(f"{donation}원 기부 감사합니다!") 
        elif int(user_input) == 4:
            print("시스템을 종료합니다")
            break

    except ValueError:
        print("1부터 4까지의 숫자만 입력해주세요")
        continue
```

```python
import random
cnt = 0
n = random.randint(1,101)       # n을 while안에 넣으면 어쩐지 매번 값이 바뀔듯..? 
box = []
while True:
    try:
        cnt += 1
        answer = int(input("1부터 100까지의 숫자를 입력해주세요~\n 기회는 열번!\n"))
        if answer < n:
            print("그거보단 커요")
            print(f"남은 도전 횟수 {10 - cnt}번")
        elif answer > n:
            print("그거보단 작아요")
            print(f"남은 도전 횟수 {10 - cnt}번")
        else:
            print(f"정답! {cnt}번만에 맞추셧어요~")
            box.append(answer)
            print(box)
            break
        if cnt == 7 and len(str(n)) > 1:
            print(f"숫자의 앞자리는 {str(n)[0]}이에요~")
        if cnt == 10:
            box.append(answer)
            print(box)
            print(f"정답은 {n}이엿습니다~")
            break
        box.append(answer)
        
    except ValueError:
            print("숫자만 적어주세요~")

            #정답을 맞추거나 기회 10번을 다 썻을때 지금까지의 헛발질을 한 이력이 나오면 좋겟어서 한번 해봣슴.
```
