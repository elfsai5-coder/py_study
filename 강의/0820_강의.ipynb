```python
import json

user = {
    'name': '김철수',
    'age': 22,
    'gender': 'male'
}

json_string = json.dumps(user,ensure_ascii = False)
print('json_string:',json_string,'type:',type(json_string))
```

```python
class MyResource:
    def __enter__(self):
        print("__enter__ 호출")
        return self
    # with 구문의 종료 시 or 예외(에러) 발생시 호출
    # 예외 발생 시 exc_type - 예외 클래스 종류
    #              exc_val - 예외 메세지
    #               exc_tb - TraceBack 메세지.
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("__exit__ 호출")
```

```python
with MyResource() as o:
    print(o)
```

```python
import csv

with open('members.csv', 'w' , encoding = 'utf-8', newline = '') as fp:
    writer = csv.writer(fp)
    writer.writerow(['이름','나이','전화번호'])
    writer.writerow(['김철수','22','010-1234-1234'])
```

```python
with open('members.csv', 'r', encoding= 'utf-8') as fp:
    reader = csv.reader(fp)
    members = list(reader)


    print(members)
```

파이썬 객체에서 전송 가능한 json 형식으로 변환 - 직렬화(serialize)
json 문자열을 dict로 변환 -> 역직렬화

```python
import asyncio
import time


async def async_work(title, sec):
    await asyncio.sleep(sec)          # 1. await 붙이기
    print(title, "완료 !")


async def main():
    await asyncio.gather(             # 2. gather에도 await 붙이기
        async_work("A작업", 2),
        async_work("B작업", 3),
        async_work("C작업", 1)
    )


start = time.perf_counter()
await main()                   # 3. asyncio.run()으로 실행
end = time.perf_counter()

print("걸린시간", end - start)
```
