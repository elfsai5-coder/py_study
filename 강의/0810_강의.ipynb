```python
def asdasd(num1,num2)
#대충 두 수의 길이는 같다 라는 항목이 있었다고 치고.
#그냥 int(num1) + int(num2)하면 안되요? 라고 할수 있지만 
num1 = 엄청 긴 수 (문자열)[::-1]
num2 = 엄청 긴 수 (문자열)[::-1]
n = 1
result = 0
for i,j in num1,num2
    i,j = int(i),int(j)
    result += (i * n) + (j * n)
    n *= 10 
return result
```
