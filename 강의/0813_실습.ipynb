```python
def message_store(massage_):
    storage = ""
    storage += massage_
    def add_message(massage):
        nonlocal storage
        storage += massage
        return storage
    return add_message

my_storage = message_store("새로운 알림이 도착했습니다")
my_storage(" 출석 체크가 완료 되었습니다.")
```

```python
def deco_border(func):
    def wrapper():
        print("========================")
        print(func())
        print("========================")
        return None
    return wrapper


@deco_border
def show_welcome():
    return "  환영합니다 스파르타님!"


show_welcome()
```

```python
def auto_tracker(func):
    def wrapper(*args, **kwargs):
        print(f"{func}함수 실행 시작 ! ")
        print(args)
        print(kwargs)       #비어있어서 {}로 표시되는듯
        result = func(*args, **kwargs)
        return result
    return wrapper


@auto_tracker
def add_numbers(a,b):
    return a + b

add_numbers(5,13)
#성공인지 아닌지 잘 모르겟다 애매..
```
