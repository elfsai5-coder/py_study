- 상속 (Inheritance)
클래스 기반의 상속 (상속에 대한 정의 -> 객체 생성 (상속 관계 성립))


class 자식 클래스 (부모 클래스):



- super() ( self 와 비슷 ? )
self는 자기 자신의 클래스를 참조. super는 부모쪽의 클래스를 참조.



부모클래스는 상위의 개념 (예를 들어서 지구)
자식클래스는 하위의 개념 (예를 들어서 각각의 국가. 한국 일본 중국 미국 등등)

```python
class Animal:
    def __init__(self, kind, 울음소리):
        self.kind = kind
        self.울음소리 = 울음소리

    def move(self):
        print(f"{self.kind} 이/가 움직이며 울부짓는다")
        print(f"{self.kind} : {self.울음소리}")

class Dog(Animal):
    def __init__(self):
        super().__init__('개','왈왈')

class Cat(Animal):
    def __init__(self):
        super().__init__('냥이','냥냥')

dog = Dog()
cat = Cat()

dog.move()
cat.move()

```

```python
class Human:                # 객체의 출처(클래스 자료형) 를 확인하는 함수 (isinstance)
    pass

    def move(self):
        print("두 발로 직립 보행 한다")

human = Human()

human.move()
```

-메서드 오버라이딩
그냥 변수 재정의 하는거랑 진배없다 ? 

```python
print("Human은 Animal 클래스로부터 생성 되었나 ? ", isinstance(Human, Animal))
print("Human은 Animal 클래스로부터 생성 되었나 ? ", isinstance(human, Human))
print("Dog은 Animal 클래스로부터 생성 되었나 ? ", isinstance(dog, Animal))
```
