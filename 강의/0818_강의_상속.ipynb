- 상속 (Inheritance)
클래스 기반의 상속 (상속에 대한 정의 -> 객체 생성 (상속 관계 성립))


class 자식 클래스 (부모 클래스):



- super() ( self 와 비슷 ? )
self는 자기 자신의 클래스를 참조. super는 부모쪽의 클래스를 참조.



부모클래스는 상위의 개념 (예를 들어서 지구)
자식클래스는 하위의 개념 (예를 들어서 각각의 국가. 한국 일본 중국 미국 등등)

```python
class Animal:
    def __init__(self, kind, 울음소리):
        self.kind = kind
        self.울음소리 = 울음소리

    def move(self):
        print(f"{self.kind} 이/가 움직이며 울부짓는다")
        print(f"{self.kind} : {self.울음소리}")

class Dog(Animal):
    def __init__(self):
        super().__init__('개','왈왈')

class Cat(Animal):
    def __init__(self):
        super().__init__('냥이','냥냥')

dog = Dog()
cat = Cat()

dog.move()
cat.move()

```

```python
class Human:                # 객체의 출처(클래스 자료형) 를 확인하는 함수 (isinstance)
    pass

    def move(self):
        print("두 발로 직립 보행 한다")

human = Human()

human.move()
```

```python
# Getter 와 Setter 매서드.

class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day

    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month

    def get_day(self):
        return self.__day

    def set_year(self, year):
        self.__year = year

    def set_month(self, month):
        if month > 12:
            month = 12
        self.__month = month

    def set_day(self, day):
        if day > 31: 
            day = 31
        self.__day = day

    def __str__(self):
        return f"year = {self.__year} month = {self.__month} day = {self.__day}"
```

```python
s1 = Schedule(2026,8,20)

s1.set_day(100)
s1.set_month(1111)


print(s1)
```
