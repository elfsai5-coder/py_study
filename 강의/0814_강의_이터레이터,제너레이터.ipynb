```python
from itertools import combinations

result = []
for i in combinations(range(10),5):
    result += [i]

result
```

-제너레이터 (Generator) : 함수의 실행을 정지하고 원할때 다시 재개 하는 형태

```python
def generate_nums():
    print("엄")
    yield 10
    print("준")            # 10을 실행하고 잠깐 정지 한다는 소린가 ? 잠깐의 기준이 뭐임 그렇다면..
    yield 20
    print("식")            # yield == 일시정지 next가 문을 다시 열어줄때까지.
    yield 30               # 어떻게 응용할수 있을까.
                            # yield = 메모리 절약. return에 비해서.
                            # 보통 return 과 yield < 둘중 하나만 씀.
```

```python
gen = generate_nums()

gen
```

```python
next(gen)
```

- 동기 / 비동기

동기 = 순서대로 실행, 결과 역시 순차적으로 도출
       // 급식실 줄서서 밥먹기
       // 서랍장에서 사탕 꺼내먹기 → 금방 끝남 (대부분의 평범한 코드)
       // 라면 끓이기 → 5분간 멍때리며 벽만 쳐다봄
       → "기다리는 동안 아무것도 못 함"

비동기 = 대기 시간을 쪼개서 다른 작업까지 처리
         // 라면 끓이면서 폰게임 하기
         // 카페 알바: 커피 내리는 동안 주문 받기
         → "기다리는 틈에 딴 일 함"

- 비동기는 파일 다운로드, API 호출, DB 쿼리처럼
  '기다림이 많은 작업'에서 활용.
  보통의 코드는 그냥 동기.

- DB 쿼리 = DB에 보내는 요청 (찾아줘/넣어줘/지워줘/고쳐줘) 해'줘'



- 제너레이터 표현식 vs 리스트 컴프리헨션

  리스트 컴프리헨션: [i for i in range(1, 10)]
  → 결과 전체를 메모리(그릇)에 한 번에 적재 (Eager)
  → [] 자체가 '그릇'이라 변수에 담아 인덱싱 가능
  → 메모리 사용 (제너레이터에 비해서)

  제너레이터 표현식: (i for i in range(1, 10))
  → 그때그때 하나씩 생성, 메모리에 안 쌓임 (Lazy)
  → ()는 '수도꼭지' — next()나 for로 틀어서 써야 함
  → sum(), min(), max()에 바로 넣으면 () 한 겹 생략 가능
     예: sum(i for i in range(1, 10))  ← 괄호 두 겹 안 써도 됨

```python
odd = (i for i in range(1,51) if i % 2 != 0)            # 제너레이터
odd2 = [i for i in range(1,51) if i % 2 != 0]           # 리스트 컴프리헨션

odd,odd2            # 1번은 만들 준비는 다됫고 명령만 내려주면 바로 만들게~!!  만들래?? 만들거면 next 버튼 눌러줘잉 << 이상태
                    # 2번은 이미 다 만들고 그릇에 담아서 제출
```
