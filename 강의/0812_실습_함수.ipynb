```python
def get_total_and_average(*n):
    if len(n) == 0:
        return "점수가 없습니다."
    return f"합계 점수:{sum(n)}  평균 점수:{sum(n)/len(n)}"


get_total_and_average(90,85,95)
```

```python
def get_average(*scores):
    if len(scores) == 0:
        return 0
    total = sum(scores)
    avg = total / len(scores)
    return avg 
    
    result = get_average() print(result)
```
