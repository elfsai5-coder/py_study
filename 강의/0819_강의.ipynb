```python
class User:
    count = 0               #  User 의 기본 count 는 0
```

```python
user1.count  = 10           # user1은 혼자서 count = 10 으로 변경

user1.count
```

```python
class Calculator:

    # cls ?  Calculator 클래스 참조 변수 ? 
    @classmethod
    def add(cls, num1, num2):   # cls 를 통해서 Class calculator 내부의 인자들에 접근가능..
        return num1 + num2
    
    @staticmethod   # 내가 class 내부에 있는건 맞는디 self는 안쓸게 
    def minus(num1,num2):
        return num1 - num2

    
```

```python
Calculator.add(30,50)
```

```python

def divide(num1, num2):
    try:
        return num1 // num2
    except ZeroDivisionError:
        print("0은 나눌수 없어요")


divide(10,0)


```

```python
def login(id,pw):
    if not (id == "user1"):
        return "아이디가 일치하지 않음"
        

    if pw != '1234':
        return "비밀번호가 틀림"
    else:
        return "로그인 성공"

print(login('usr1','124'))
```

```python

def login(id, pw):
    try:
        if id != 'user1':
            raise ValueError                    # 이걸 실제로 한다면 user1 이랑 1234는 따로 저장해놓은 db에서 id,pw와 마찬가지로 변수로 삽입후 검증하는게 좋아보임.
        if pw != '1234':
            raise ValueError
        else:
            return "로그인 완료"
    except ValueError:
        print("아이디 혹은 비밀번호를 체크")
    


login('user1','1234')


# 예를 들어서 user_input_id, user_input_pw 변수를 각각 만들고 아이디 비번 인풋을 받음.
# 그 후 login(db_id, db_pw, user_input_id, user_input_pw): 함수에 넣고 각각 검증

```
