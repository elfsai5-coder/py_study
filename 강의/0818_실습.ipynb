```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def info(self):
        return f"책 제목 : {self.title} 저자 : {self.author}"
    
어린_왕자 = Book("어린 왕자",'생택쥐페리')
나의_라임_오렌지_나무 = Book("나의 라임 오렌지 나무", "J.M 바스콘셀루스")


print(어린_왕자.info())
print(나의_라임_오렌지_나무.info())
```

```python
class SmartLamp:
    def __init__(self, model):
        self.model = model
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print(f"{self.model} 조명이 켜졋습니다.")
        return
    
    def turn_off(self):
        self.is_on = False
        print(f"{self.model} 조명이 꺼졋습니다.")

    def turn_on_off(self):
        if self.is_on == True:
            self.is_on = False
            print(f"{self.model}의 조명이 꺼졋습니다.")
            return
        else:
            self.is_on = True
            print(f"{self.model}의 조명이 켜졋습니다.")
            return

geene = SmartLamp('지니')
geene.turn_on()
geene.turn_off()
geene.turn_on_off()
geene.turn_on_off()
geene.turn_on_off()
geene.turn_on_off()
geene.turn_on_off()


```

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):
        self.hp = self.hp - damage
        if self.hp <= 0:
            self.hp = 0
        print(f"남은 HP : {self.hp}")
        if self.hp == 0:
            print(f"{self.name} 캐릭터가 쓰러졋습니다.")


zi존전사 = Hero("zi존전사")


zi존전사.take_damage(30)
zi존전사.take_damage(30)
zi존전사.take_damage(30)
zi존전사.take_damage(30)

```

```python
class Phone:
    def call(self):
        return "통화를 시작합니다"

class Smartphone(Phone):
    def call(self):
        return "영상 통화를 시작합니다."

    def search_wep(self):
        return "인터넷 검색을 시작합니다."


samsumg = Smartphone()

print(samsumg.call())
samsumg.search_wep()
```

```python
class User:
    def __init__(self, name):
        self.name = name


class VIPUser(User):
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point

    def __str__(self):
        return f" 이름 : {self.name} , 포인트 : {self.point}"

이순자 = VIPUser("이순자", 1500)

print(이순자)
```
