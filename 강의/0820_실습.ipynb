```python
import time

def download_page(sec):
    start = time.perf_counter()
    time.sleep(sec)
    time.sleep(sec)
    end = time.perf_counter()
    print(f" 작업 완료 ! 총 소요 시간 {end - start}")
    return 


download_page(2)
```

```python
import asyncio
import time

async def download_page(title,sec):
    print("작업 시작 ! ")
    await asyncio.sleep(sec)
    print(title, "완료 !")


async def main():
    start = time.perf_counter()
    await asyncio.gather(
        download_page("메인페이지",2),
        download_page("상세페이지",2)
    )
    end = time.perf_counter()
    print(f" 작업 완료 ! 총 소요 시간 {end - start}")



await main()


```

```python
import asyncio
import time

async def fetch_weather(city,sec):
    print(f"{city}의 날씨 데이터를 수집합니다 ...")
    await asyncio.sleep(sec)
    print(f"{city}의 날씨 데이터를 수집 완료 하엿습니다.")


async def main():
    start = time.perf_counter()
    await asyncio.gather(
        fetch_weather('뉴욕',2.5),
        fetch_weather("런던",1),
        fetch_weather('서울', 1.5)
    )
    end = time.perf_counter()
    print(f" 모든 도시의 날씨 데이터를 수집 완료 하엿습니다. 총 소요시간 : {end - start} 초 ")



await main()


```
