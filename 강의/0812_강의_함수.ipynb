```python
def avg(*score):
    if len(score) == 0:
        return 0
    result = sum(score) // len(score)
    return result

avg(11,33,44,55,11,22,23,49,85,92,12)
```

```python
def print_user_info(**info):    # 키워드 가변 인수. //딕셔너리 형식으로 만들어주는 거 // kwargs라고 함. 별 두개 붙히면 됨.
    print(info)
```

```python
print_user_info(name = '찰스', age = 20)
```

```python
def print_user_info(name,age,*args,**kwargs):
    print(f"name: {name}")
    print(f"age: {age}")
    print(f"{name}의 하루 공부 시간: {args}")
    print(f"{name}의 각종 세부 정보")
    print(kwargs)
```

```python
def calc(n1,n2):                                # 튜플 or 리스트로 묶으면 내부 원소는 여러개지만 포장해서 한개 처리
    return [n1 + n2 , n1 - n2 , n1 * n2 , n1 // n2]
```
