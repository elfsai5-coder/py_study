- 전역변수 (global) : 함수 외부에 선언되는 변수들. 코드 전역에서 접근 가능하다 ( 함수 내부에선 global 변수명 으로 접근)
- 지역변수 (local) : 함수 내부에서 선언되는 변수들. 약간 깡촌같은 느낌. 도시(global)로 나가면 아무도 알아보는 사람이 없음 ..

```python
box = []
def solution(n):
    for i in range(n):
        box.append(i)
    
    return box  

solution(5)     # 엥 왜 되냐. 저번에 안됫는디. 저번에 안되가지고  global 변수명 으로 해결했었는데.. 아무튼 되니깐. 이렇게 하면 됨..
```

```python
box = []
def solution(n):
    for i in range(n):
        box = [100000]          # 함수 (local) 안에 새롭게 box = [100000]을 넣어봄
        box.append(i)
    
    return box

solution(5)                     # 글로벌보다 로컬이 우선된다. 
```

L ( 로컬 ) -> E (Enclosing (?) 외부 함수 변수) -> G (글로벌) -> B( Builtin ) 내장 함수

```python
a = '엄'
print(a)
def outer():
    a = '준'
    print(a)
    def inner():
        a = '식'
        print(a)
    return inner()
outer()
```

```python
a = '엄'
print(a)
def outer():
    a = '준'
    print(a)
    def inner():
        print(a)
    return inner()          # inner 쪽의 a의 정의를 지우자 '엄 준 준' 이 되버렷다.
outer()
```

```python
a = '엄'
print(a)
def outer():
    a = '준'
    print(a)
    def inner():
        global a                # inner 쪽의 a의 정의를 지우고 global에 있는 a의 정의를 사용하라고 하니 "엄 준 엄"이 되버렷다.
        print(a)
    return inner()
outer()
```

```python
def outer(num1):
    def inner():                # inner의 변수를 multiply의 변수로 지정하는 방법을 모르겟다.
        num2 = num1 * 3         # 그래서 그냥 inner를 비워두고 내부에서 num2의 값을 지정해서 multiply를 실행해보니 우선 작동함
        return num1 + num2      # 근데 이러면 inner의 존재 이의가 뭐지 ..? 그냥 outer만 있어도 실행엔 전혀 지장이 없는데.
    return inner()

def multiply(n):
    return n * 5

multiply(outer(3))

```

```python
def multiplier(factor):
    def multiply(n):
        return n * factor
    return multiply # ()를 떼고 return 하면 그 값을 보존한다? 

double = multiplier(2)          # multiplier에,아니 그니깐 정확히는  multiplier의 내부 multiply에 return multiply로 함수가 끝나니깐.
triple = multiplier(3)          # 거기에 double, triple의 숫자를 보존시켜놓은건가 ? 

print(double(20))               # double(20) 의 (20) 은 multiplier의 변수로 20이 박히는건가? 
print(triple(30))               # 이해 했다 ㅇㅇ ㅇㅇ ㅇㅇ ㅇㅇㅇㅇㅇ
                                # double = multiplier(2) // 라는 변수 정의 앞으로 double은 따로 바뀌는 일이 없는한 무조건 이거인거임..
                                # double(20) 을 돌리면 factor = 2인 double 함수에 n = 20으로 들어감.
                                # 내부에선 값을 뱉긴하짐나 표시해주지는 않으므로 print(double(20)) << 이 필수적임
```

```python
multiplier(2)(20) # inner 함수의 값은 이렇게 뒤에 괄호치고 달라붙는다.
```
