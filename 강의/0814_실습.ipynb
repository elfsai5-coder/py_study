```python
def word():             # 우선 word() 실행. 그 후 print("대박")
    print("대박")
    yield  10             # 여기서 next(word())를 하면 print("사건") 으로 넘어가야 하는거 아닌가 ? 
    print("사건")
    yield 20
    print("성공")
    yield 30 



result = word()        # 왜 result 에 word()를 담고 실행하니까 되는거지 ? 
                        # result = 끊어지지 않는 id ?를 가진 변수라서 쭉 이어져서 근가 ? 
                        # 쌩으로 word()를 하면 "대박 대박 대박" << 첫  yield에서 초기화 된다는 소리. id가 매번 바뀐다 ? 
print(next(result))
print(next(result))
print(next(result))


```

```python
from datetime import datetime

def today():                        # 이유는 몰겟는데 datetime.now는 에러가 남
    now = datetime.now()      #  datetime.now 전체가 함수명인가..?
    data = list(map(str,[now.year, now.month, now.day]))
    result = []
    for i in data:
        if len(i) == 1:
            i += '0'
            i = i[::-1]
        result.append(i)
    return '-'.join(result)          # 이거 이렇게 복잡하게 할거 아닌거같은데...

today()
```
