```python
# 클래스 == 객체가 탄생? 하면서 담고있을 기본적인 정보들의 규격을 제시하는 객체 공장 느낌 (더 정확히는 클래스가 아닌 __init__ ? )
# 객체 == 서로 다른 정보를 보유중인 서로 다른 독립된 개체
```

```python
import random
class Passenger:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.gender = 'male' if random.randint(0,1) == 1 else 'female'

    def __str__(self):
        return f" 이름 : {self.name}, 나이: {self.age}세, 성별 : {self.gender}"

    def __repr__(self):
        return f"Passenger(name='{self.name}', age = {self.age}, 성별 = {self.gender})"


class Bus:
    def __init__(self, number):
        self.number = number
        self.count = 0

    def ride(self):
        print("탑승")
        self.count += random.randint(0,15)
        if self.count >= 50:
            print("더는 탑승 할 수 없습니다")
            self.count = 50
        print(f"{self.number}번 버스의 현재 탑승객 수 : {self.count}")
        return 

    def off(self):
        print("하차")
        self.count -= random.randint(0,15)
        if self.count < 0:
            self.count = 0
        print(f"{self.number}번 버스의 현재 탑승객 수 : {self.count}")
        return

    def __str__(self):
        return f"승객 수 : {self.count} " 

    def __repr__(self):
        return "REPR"
        
```

```python
bus_1 = Bus("501")
bus_2 = Bus("5515B")

bus_2.ride()            # 5515B 버스와 501버스가 서로 다른 객체임을 확인 할수 있다 (서로 승객수의 카운트가 개별적으로 적용 ! )
bus_1.ride()
bus_2.ride()
bus_1.ride()
print(bus_1)
```
