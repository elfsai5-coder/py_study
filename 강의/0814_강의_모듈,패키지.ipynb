```python
import itertools as it # 앞으로는 이터툴을 it라고 부르도록 하마.
result = []
for i in it.combinations(range(1,6),3):     # itertools가 아닌 it.으로도 combinations가 불러지는 모습
    if sum(i) > 8:
        result += [i]

print(result)
f"총 : {len(result)} 개"
```
