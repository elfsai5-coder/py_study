```python
def factorial1(n):
    result = 1
    for i in range(1,n+1):
        result *= i
    return result



factorial1(10)

```

```python
import time

def timer(callback,n):      # 이런식으로 돌려쓰기 좋은 유용한 extend같은 함수들을 데코레이터 라고 하는듯하다.
    start = time.time()

    result = callback(n)

    end = time.time()
    print("걸린 시간:", end - start)
    return result
```

```python
def timer(callback):                
    def wrapper(n):
        start = time.time() # 공통 기능
        result = callback(n)    # 핵심 기능
        end = time.time()
        print("걸린시간:",end - start)
        return result
    return wrapper


@timer
def factorial1(n):
    result = 1
    for i in range(1,n+1):
        result *= i
    return result



factorial1(10)
```
