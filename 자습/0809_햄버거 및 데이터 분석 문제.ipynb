어제 했던 햄버거 문제에 대해 어젯밤 잠들기 직전에 다른 방식이 생각나서 구현해봣다.
재귀방식을 써서 효율이 좋지않지만 우선 오류없이(시간초과빼고) 작동은 하는듯하다

```python
def solution(ingredient):
    for i in range(len(ingredient) - 3):                            # 아래 if문에서 i+4까지 한번에 체크하므로 range는 3만큼 줄여서
        if ingredient[i:i+4] == [1,2,3,1]:                          # if문에서 조건이 맞으면 i:i+4를 제외한 부분을 다시 함수 호출
            return 1 + solution(ingredient[:i] + ingredient[i+4:])  # 조건을 못찾으면 0을 return
    return 0
```

```python
def solution(ingredient):               # 어제했던거랑 똑같은 스택형인데 while을 써본 방식
    box,cnt = [],0                      # 함수 호출도 한번에 끝나고 가독성도 좋음
    for i in ingredient:
        box.append(i)
        while box[-4:] == [1,2,3,1]:
            cnt += 1
            del box[-4:]
    return cnt
```

AI 엔지니어인 현식이는 데이터를 분석하는 작업을 진행하고 있습니다. 데이터는 ["코드 번호(code)", "제조일(date)", "최대 수량(maximum)", "현재 수량(remain)"]으로 구성되어 있으며 현식이는 이 데이터들 중 조건을 만족하는 데이터만 뽑아서 정렬하려 합니다.

예를 들어 다음과 같이 데이터가 주어진다면

data = [[1, 20300104, 100, 80], [2, 20300804, 847, 37], [3, 20300401, 10, 8]]
이 데이터는 다음 표처럼 나타낼 수 있습니다.

code	date	maximum	remain
1	20300104	100	80
2	20300804	847	37
3	20300401	10	8
주어진 데이터 중 "제조일이 20300501 이전인 물건들을 현재 수량이 적은 순서"로 정렬해야 한다면 조건에 맞게 가공된 데이터는 다음과 같습니다.

data = [[3,20300401,10,8],[1,20300104,100,80]]
정렬한 데이터들이 담긴 이차원 정수 리스트 data와 어떤 정보를 기준으로 데이터를 뽑아낼지를 의미하는 문자열 ext, 뽑아낼 정보의 기준값을 나타내는 정수 val_ext, 정보를 정렬할 기준이 되는 문자열 sort_by가 주어집니다.

data에서 ext 값이 val_ext보다 작은 데이터만 뽑은 후, sort_by에 해당하는 값을 기준으로 오름차순으로 정렬하여 return 하도록 solution 함수를 완성해 주세요. 단, 조건을 만족하는 데이터는 항상 한 개 이상 존재합니다.

```python
def solution(data, ext, val_ext, sort_by):
    result = []
    if ext == 'code':
        ext = 0
    elif ext == 'date':                             # ext, sort_by의 값을 정수(인덱스)화
        ext = 1
    elif ext == 'maximum':
        ext = 2
    else:
        ext = 3
        
    if sort_by == 'code':
        sort_by = 0
    elif sort_by == 'date':
        sort_by = 1
    elif sort_by == 'maximum':
        sort_by = 2
    else:
        sort_by = 3
    for i in data:
        if i[ext] < val_ext:                        # val_ext 보다 작은 값만 result에 append 
            result.append(i)
    result.sort(key = lambda x: x[sort_by])         # 그 후 sort_by의 값을 기준으로 정렬
    return result
```

```python
def solution(엄준식, 준, 식, 준식):                                 #다른 사람의 풀이. 해석이 빡세다
    어떻게사람이름이 = ["code","date","maximum","remain"]            # ext,sort_by의 정수변환을 한줄에 짧게
    엄준식 = [엄 for 엄 in 엄준식 if 엄[어떻게사람이름이.index(준)]<식] # 엄준식 == data 니깐,i를 꺼내서 리스트 컴프리헨션으로 간단하게 거르고
    엄준식.sort(key=lambda 엄: 엄[어떻게사람이름이.index(준식)])       # 정렬. result를 만들지 않고 간결하고 이해하기 쉽게 (변수명이 이해하기 어렵긴 한데) 풀어낸게 인상적
    return 엄준식
```

```python
def solution(data, ext, val_ext, sort_by):

    if ext == "code": newExt = 0            #if문이 한줄로도 가능하다는걸 첨봐서 
    elif ext == "date": newExt = 1          #단 하나의 문장만 실행할 경우엔 성립한다고 함
    elif ext == "maximum": newExt = 2       #남용할시 읽기 지저분해져서 피하는 편이 좋다고
    else: newExt = 3

    if sort_by == "code": newSort = 0
    elif sort_by == "date": newSort = 1
    elif sort_by == "maximum": newSort = 2
    else: newSort = 3

    answer = []
    for d in data:
        if d[newExt] < val_ext:
            answer.append(d)

    return sorted(answer, key=lambda x: x[newSort])
```
