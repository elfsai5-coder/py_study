JadenCase란 모든 단어의 첫 문자가 대문자이고, 그 외의 알파벳은 소문자인 문자열입니다. 단, 첫 문자가 알파벳이 아닐 때에는 이어지는 알파벳은 소문자로 쓰면 됩니다. (첫 번째 입출력 예 참고)
문자열 s가 주어졌을 때, s를 JadenCase로 바꾼 문자열을 리턴하는 함수, solution을 완성해주세요.

제한 조건
s는 길이 1 이상 200 이하인 문자열입니다.
s는 알파벳과 숫자, 공백문자(" ")로 이루어져 있습니다.
숫자는 단어의 첫 문자로만 나옵니다.
숫자로만 이루어진 단어는 없습니다.
공백문자가 연속해서 나올 수 있습니다.

```python
def solution(s):                                    # 일단 코드 자체도 좀 누덕누덕 하고 공백문자가 연속으로 나오는 부분에서 에러 발생 하는듯
    s = s.split(" ")                                # join 할때 전부다 공백 한칸으로 취급하고 있기 때문에.
    result = []                                     # 그래서 그 부분을 해결 해야함..
    for i in s:                                     # 전에 이거 비슷한거 한번 헀었던거같은데 기억이 날랑 말랑 함 흠 
        row = ""                                    # 애초에 split할때 원본이 소실되는게 좀 그건디.
        for j in i:                                 # 
            if j == i[0] and i[0].isalpha():
                row += j.upper()
            elif j == i[0] and i[0].isdigit():
                row += j
            if j != i[0] and j.isalpha():
                row += j.lower()
        result.append(row)
    return ' '.join(result)
```

0과 1로 이루어진 어떤 문자열 x에 대한 이진 변환을 다음과 같이 정의합니다.

x의 모든 0을 제거합니다.
x의 길이를 c라고 하면, x를 "c를 2진법으로 표현한 문자열"로 바꿉니다.
예를 들어, x = "0111010"이라면, x에 이진 변환을 가하면 x = "0111010" -> "1111" -> "100" 이 됩니다.

0과 1로 이루어진 문자열 s가 매개변수로 주어집니다. s가 "1"이 될 때까지 계속해서 s에 이진 변환을 가했을 때, 이진 변환의 횟수와 변환 과정에서 제거된 모든 0의 개수를 각각 배열에 담아 return 하도록 solution 함수를 완성해주세요.

```python
def solution(s):                    # 지문이 무슨 소리인지 이해하는데만 10분걸림
    cnt = 0
    removed = 0
    while s != '1':                 # s가 '1'이 될때까지 반복
        removed += s.count('0')     # removed에 지울 '0'의 갯수를 추가후 삭제
        s = s.replace('0','')
        s = bin(len(s))[2:]         # len(s)를 2진수로 변경 후 제일 앞에 2글자는 슬라이싱 // ob어쩌구 하는 글자가 나옴
        cnt += 1
    return [cnt,removed]
```
