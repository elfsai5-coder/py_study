```python
def stairs(n):
    if n == 1:
        return 1
    if n == 2:
        return 2
    return stairs(n - 1) + stairs(n - 2)

stairs(30)
```

```python
def stairs(n):
    if n == 1:
        return 1
    if n == 2:
        return 2
    cnt_1, cnt_2 = 1,2
    for _ in range(3, n + 1): # n이 1일때랑 2일때는 이미 저 위에서 셋으므로 3부터 시작.
        temp = cnt_2   #cnt_2의 값을 임시 변수 temp에 잠깐 저장 (아래서 써야함)
        cnt_2 = cnt_1  +cnt_2
        cnt_1 = temp
    return cnt_2
stairs(60)
```

```python
# 어떤 수 x와 자연수 n이 주어질 때, x부터 시작해 x씩 증가하는 숫자를 n개 지니는 리스트를 반환하는 함수 sequence(x, n)을 작성
def sequence(x, n):
    return [x * i for i in range(1,n+1)]

sequence(3,20)
```
