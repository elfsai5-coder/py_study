문제 설명
JadenCase란 모든 단어의 첫 문자가 대문자이고, 그 외의 알파벳은 소문자인 문자열입니다. 단, 첫 문자가 알파벳이 아닐 때에는 이어지는 알파벳은 소문자로 쓰면 됩니다. (첫 번째 입출력 예 참고)
문자열 s가 주어졌을 때, s를 JadenCase로 바꾼 문자열을 리턴하는 함수, solution을 완성해주세요.

제한 조건
s는 길이 1 이상 200 이하인 문자열입니다.
s는 알파벳과 숫자, 공백문자(" ")로 이루어져 있습니다.
숫자는 단어의 첫 문자로만 나옵니다.
숫자로만 이루어진 단어는 없습니다.
공백문자가 연속해서 나올 수 있습니다.

```python
def solution(s):                                    # 일단 코드 자체도 좀 누덕누덕 하고 공백문자가 연속으로 나오는 부분에서 에러 발생 하는듯
    s = s.split(" ")                                # join 할때 전부다 공백 한칸으로 취급하고 있기 때문에.
    result = []                                     # 그래서 그 부분을 해결 해야함..
    for i in s:                                     # 전에 이거 비슷한거 한번 헀었던거같은데 기억이 날랑 말랑 함 흠 
        row = ""                                    # 애초에 split할때 원본이 소실되는게 좀 그건디.
        for j in i:                                 # 
            if j == i[0] and i[0].isalpha():
                row += j.upper()
            elif j == i[0] and i[0].isdigit():
                row += j
            if j != i[0] and j.isalpha():
                row += j.lower()
        result.append(row)
    return ' '.join(result)
```
