```python
def apply_tax(rate):
    def calculate(price):
        return price * ( 1 + rate)
    return calculate

vat = apply_tax(0.1)
luxury_vat = apply_tax(0.2)

print(vat(50000))
print(luxury_vat(50000))
```

```python
def announce(func):
    def wrapper():
        print("시작")
        func()          # 실행할 함수
        print("끝")
    return wrapper


@announce
def hello():
    print("안녕")



hello()
```

```python
import itertools
result = []

for i in itertools.combinations(range(1,6),3):
    if sum(i) > 8:
        result.append(i)

print(len(result),result)
```

```python
def odd_nums():
    n = 1
    while True:
        yield n
        n += 2

gen = odd_nums()

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```
