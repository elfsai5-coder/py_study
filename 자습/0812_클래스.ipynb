```python
class Cookie:
    def __init__(self, flavor):
        self.flavor = flavor

    def describe(self):
        return f"이 쿠키는 {self.flavor}맛이랍니다."
```

__init__ : 생성자 라고 부른다고 한다.쿠키를 굽는순간 자동으로 발동되며..재료(인자)를 받아들인다.

self : 자기 자신을 가리키는 표지판. 첫 인자에 항상 붙어있어야 하며,인스턴트의 상태를 저장,혹은 접근할때 사용한다.

아래에 한번 예문을 연습해보자. ( 따라 적으면서 그냥 손으로 머리로 그 느낌을 좀 어.)

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner   # self.owner = owner  self 가 자기자신을 가리키는거니깐. 아무튼 자기는 owner이다. 뭐 이런건가.
        self.balance = balance      

    def deposit(self, amount):
        self.balance += amount
        return f"{self.owner}님. {amount}원 입금 되었어요."

    def withdraw(self, amount):
        if amount > self.balance:
            return f"잔액이 부족해요 {self.owner}님"
        else:
            self.balance -= amount
            return f"{amount}원 출금하엿습니다. 잔액 {self.balance}원"

account_1 = BankAccount("홍길동",100000)
account_2 = BankAccount("장영실",500)
print(account_1.deposit(50000))            # class 내부의 def 함수들은 약간 .append() 처럼 뒤에 . 찍고 실행 되나 보다.
print(account_1.withdraw(150000))          # 그러니깐, self가 붙은 객체들을 따로따로 관리해서..그러는듯한 느낌
print(account_2.withdraw(10000))
```

```python
class Person:
    def __init__(self, name, age): # 아하 딱알았다. 그러니까 __init__의 역할은 클래스에서 앞으로 사용할 변수(기초 골격)를 정의하는것?
        self.name = name
        self.age = age

    def introduce(self):            # 내부가 self로 끝나는거면..이 함수는 그냥 비어있는건가? 
        return f"안녕하세요 저는 {self.name}입니다. {self.age}살이에요."

    def birth_day(self):
        self.age += 1
        return f"생일축하! 이제 {self.age}살 이에요"

p1 = Person("홍길동",30)            # 아하 내부에 self만 있는 함수는 추가적으로 값을 받아들일 필요가 없는. 함수이다.(기존 변수들만으로 충분한)
p2 = Person("장영실",25)

print(p1.introduce())
print(p2.introduce())
print(p2.birth_day())
print(p1.introduce())
print(p2.introduce())
```

- 인스턴스 변수 ( 객체마다 가지고 있는 고유의 값. 예를들어 이름. 나이)
- 클래스 변수 (객체들이 서로 공유할수 있는 변수. 예를들어 장영실과 홍길동이 사는 집.)

```python
import random
class Cookie:
    def __init__(self, flavor):
        self.flavor = flavor
        cnt = 0 #이거 되는건가 ? __init__이 약간 본진. 커맨드센터 느낌이니깐 되지 않을까 ?

    def make_cookie(self):
        random_cookie = random.randint(1,6)
        cnt += random_cookie
        return f"얏호! {self.flavor}맛 쿠키가 {random_cookie}명 늘었다!"

banana = Cookie("Banana")
print(banana.make_cookie())
print(banana.make_cookie())
print(banana.make_cookie())
print(banana.make_cookie())

```

cnt를 init의 내부에 두니 작동이 안된다.

```python
import random
class Cookie:
    baked_count = 0
    def __init__(self, flavor):
        self.flavor = flavor
        Cookie.baked_count += 1

    def make_cookie(self):
        random_cookie = random.randint(1,6)
        Cookie.baked_count += random_cookie
        return f"얏호! {self.flavor}맛 쿠키가 {random_cookie}명 늘었다! 총{Cookie.baked_count}명 ! "
        # Cookie.baked_count = Cookie 클래스에 귀속된 baked_count 라는 변수를 칭하는 말. 클래스 변수

banana = Cookie("Banana")       # 그리고 바나나맛 쿠키는 인스턴스 변수
print(banana.make_cookie())
print(banana.make_cookie())
print(banana.make_cookie())
print(banana.make_cookie())

```

```python
class Library:
    total_books = 0
    def __init__(self, name, books):        # 아 잠깐만 방금 Library.total_books += self.books 적으면서 감을 딱 잡아버림
        self.name = name                    # __init__은 출생신고서 같은거인거임..
        self.books = books                  # 이 응애가 태어날때 이런이런 사항들을 실행하도록해! 하는
        Library.total_books += self.books
    
    def add_books(self, add):
        Library.total_books += add
        self.books += add
        print(f"우리나라 도서관의 총 소장권수 : {Library.total_books}권")
        return f"{self.name} 도서관에 책이 {add}권 추가되었어요. {self.name} 도서관의 총 소장 권수 : {self.books} 권"

서울 = Library("서울",0)
부산 = Library("부산",150)
제주 = Library("제주",293)

print(서울.add_books(1))
print(제주.add_books(82))
print(부산.add_books(38))
print(서울.add_books(1))
```

- init은 출생신고서.
- 인스턴스 변수는 객체마다 고유의 어쩌고
- 클래스 변수는 모두가 공유하는 거
